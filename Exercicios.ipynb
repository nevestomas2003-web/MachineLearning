```python

#Exercicio 3

list_1=[1,1,1,1]
lista_numeros = []

print("Vamos criar uma lista de números! Escreve 'fim' quando quiseres parar.")

# 2. Iniciar um ciclo infinito (ele vai repetir até bater no 'break')
while True:
    # 3. Pedir algo ao utilizador
    resposta = input("Escreve um número: ")

      # 4. Condição de paragem: se a pessoa escrever 'fim', o ciclo pára
    if resposta.lower() == "fim" or resposta == " ":
      break


    # 5. Se não escreveu 'fim', então assumimos que é um número.
    # Transformamos o texto em número inteiro e usamos o append!
    numero_convertido = int(resposta)
    lista_numeros.append(numero_convertido)

# 6. No final, quando o ciclo termina, mostramos o resultado
print("A tua lista final ficou assim:", lista_numeros)
```
