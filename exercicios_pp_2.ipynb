```python
import numpy as np

# 1. Crias as tuas 3 folhas separadamente
folha_uns = np.ones((3, 3))
folha_zeros = np.zeros((3, 3))
folha_diagonal = np.diag([1, 2, 3])

# 2. Juntas tudo no mesmo "ficheiro" (o tal array 3D)
array_final = np.array([folha_uns, folha_zeros, folha_diagonal])

print(array_final)
```

```python
import numpy as np

a = np.arange(0, 7)

b = np.linspace(0, 6, 7)

c=np.diag((1,1,1,1))
print("escreve o a\n",a, "escreve o b\n", b,
      "\n depois o c\n",c)

rng = np.random.default_rng()

# 2. Criar a folha de Excel 5x5 com números entre [0, 1[
# Nota: passamos um tuple (5, 5) para avisar que queremos 5 linhas e 5 colunas
matriz = rng.random((5, 5))

# 3. Arredondar tudo a 2 casas decimais com a ferramenta np.round()
matriz_arredondada = np.round(matriz, 2)

# 4. Encontrar o valor mínimo e máximo
valor_minimo = matriz_arredondada.min()
valor_maximo = matriz_arredondada.max()

# Mostrar os resultados no ecrã
print("Matriz 5x5:\n", matriz_arredondada)
print("\nMínimo:", valor_minimo)
print("Máximo:", valor_maximo)
```
