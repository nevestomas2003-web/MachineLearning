```python
#Exercicio 2 -  Ask the user for a word and count how many vowels it has.
# input() : data input by the user

vowels="aeiou"
user_text=input("Diz me uma palavra ou uma frase?")
count=0

for i in user_text:
  if i in vowels:
    count=count+1;

print("o total de vogais é:", count)
```

```python
#Exercise 3 - Create two lists, one with even numbers and another with odd numbers, based on a string of digits entered by the user (e.g. 1234)

num = input(" input a sequence of even and odd digits")
evenList=[]
oddList=[]

for i in num:
  if (int(i) %2 == 0):
    evenList.append(i)
  else:
    oddList.append(i)

print("Even list", evenList)
print("Odd list", oddList)

```

```python
#exercise 4 Create a filter to identify names with less than 4 letters. The user must enter 5 names

namesUnder4letters = []
namesOver4letters = []

for i in range(5):
    name = input("Escreve um nome: ")


    if len(name) < 4:
        namesUnder4letters.append(name)
    else:
        namesOver4letters.append(name)



print("Nomes com menos de 4 letras:", namesUnder4letters)
print("Nomes com mais de 4 letras:", namesOver4letters)
```

```python
#Based on List Comprehensions calculate the squares only for the even numbers of nums
#▪ nums= [0,1,2,3,4,5,6,7]


a = [0,1,2,3,4,5,6,7]

B = [i**2 for i in a if i%2==0]

print(B)
```

```python
#Create a dictionary comprehension based on num=[0,1,2,3,4,5,6], such that:
#▪ the key is a number and the value the respective square but only if the original number is
#higher than 2

num=[0,1,2,3,4,5,6]

d={i:i**2 for i in num if i > 2}

print(d)
```

```python
#Consider a list made up of sublists.
#▪ l=[[10, 20, 30],[40, 50],[200,300,400]]
#▪ Using a list comprehension, calculate the sum of the values of the inner lists.

l=[[10, 20, 30],[40, 50],[200,300,400]]

sum_list=[sum(i) for i in l]

print(sum_list)
```

```python
class Student(object):
  def __init__(self,name, age, gender):
    self.name=name
    self.age=age
    self.gender=gender
  def SetAge(self):
    print("age:", self.age)
  def SetName(self):
    print("name:", self.name)
  def SetGender(self):
    print("gender:", self.gender)




class MasterStudent(Student):
  intership="from march to july"

newMasterStudent=MasterStudent("tomás", "masculino",22)

newMasterStudent.SetAge()
newMasterStudent.SetName()
newMasterStudent.SetGender()
print("intership" , newMasterStudent.intership)
```
